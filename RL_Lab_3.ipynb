```python
import random

# Part A: Machine initialization
machines = ["A", "B", "C"]

# N = number of plays, R = total reward, Q = estimated average reward
N = {machine: 0 for machine in machines}
R = {machine: 0.0 for machine in machines}
Q = {machine: 0.0 for machine in machines}

epsilon = 0.2
total_reward = 0.0

print("Modified Epsilon-Greedy Multi-Armed Bandit")
print("Number of plays: 10")
print("Epsilon:", epsilon)
print("\nSuggested reward patterns:")
print("A: 3, 4, 2, 5")
print("B: 6, 7, 5, 8")
print("C: 4, 3, 5, 4")

for play in range(1, 11):
    # Find a machine with the highest Q-value.
    # random.choice handles ties fairly when all Q-values are equal.
    highest_q = max(Q.values())
    best_machines = [m for m in machines if Q[m] == highest_q]
    best_machine = random.choice(best_machines)

    r = random.random()

    # Part C: Modified epsilon-greedy decision
    if r < epsilon:
        # Exploration: choose any machine except the current best machine
        other_machines = [m for m in machines if m != best_machine]
        selected_machine = random.choice(other_machines)
        decision = "Exploration"
    else:
        # Exploitation: choose the machine with highest Q-value
        selected_machine = best_machine
        decision = "Exploitation"

    print(f"\nPlay {play}: {decision}")
    print("Selected machine:", selected_machine)

    # Part D: Reward collection
    reward = float(input("Enter reward: "))

    # Part E: Incremental sample-average Q update
    N[selected_machine] += 1
    R[selected_machine] += reward

    Q[selected_machine] = Q[selected_machine] + (
        reward - Q[selected_machine]
    ) / N[selected_machine]

    total_reward += reward

# Part F: Display final tables
print("\n--- Final N, R, Q Tables ---")
print("Machine\tN (plays)\tR (total reward)\tQ (average reward)")

for machine in machines:
    print(f"{machine}\t{N[machine]}\t\t{R[machine]:.2f}\t\t\t{Q[machine]:.2f}")

print("\nTotal reward received by the player:", total_reward)
```
